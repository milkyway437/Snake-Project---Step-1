# بررسی محیط کاری پروژه (check_env)

این نوت‌بوک اطلاعات مربوط به محیط اجرای پایتون و کتابخانه‌های مورد استفاده در پروژه را ثبت می‌کند.

```python
import sys
import platform

print("=== Python Version ===")
print(f"Version       : {platform.python_version()}")
print(f"Full Version  : {sys.version}")
print(f"Executable    : {sys.executable}")
```

```python
import numpy as np

print("=== NumPy ===")
print(f"Version       : {np.__version__}")
print(f"Install Path  : {np.__file__}")
```

```python
import matplotlib

print("=== Matplotlib ===")
print(f"Version       : {matplotlib.__version__}")
print(f"Install Path  : {matplotlib.__file__}")
```

```python
import sys
import platform
import matplotlib
import numpy as np

print("=" * 40)
print("خلاصه محیط کاری پروژه")
print("=" * 40)
print(f"Python      : {platform.python_version()}")
print(f"Executable  : {sys.executable}")
print(f"NumPy       : {np.__version__}")
print(f"Matplotlib  : {matplotlib.__version__}")
```
